```python
#install libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#get top feature data
url = "https://raw.githubusercontent.com/austinmallie/ADS599_Capstone/main/Models/Feature%20Importance/feature_importance_xgb.csv"
df = pd.read_csv(url)
df.shape
```

```python
df.head(10)
```

```python
#all features
df_sorted = df.sort_values(by='Importance', ascending=False)
plt.figure(figsize=(12, 15))
sns.barplot(
    x='Importance',
    y='Feature',
    data=df_sorted,
    palette='viridis'
)

plt.title('XGBoost Feature Importance (All Features)', fontsize=16)
plt.xlabel('Importance Score', fontsize=12)
plt.ylabel('Features', fontsize=10)

plt.tight_layout()

plt.show()
```

```python
#top 10 only

df_sorted = df.sort_values(by='Importance', ascending=False)
df_top10 = df_sorted.head(10)
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

plot = sns.barplot(
    x='Importance',
    y='Feature',
    data=df_top10,
    palette='viridis'
)
plt.title('Top 10 Most Important Features (XGBoost)', fontsize=16, pad=20)
plt.xlabel('Importance Score (Gain)', fontsize=12)
plt.ylabel('') # Removing the 'Features' label for a cleaner look

for i in plot.containers:
    plot.bar_label(i, padding=3, fmt='%.4f')

plt.tight_layout()
plt.show()
```
